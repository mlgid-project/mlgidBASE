# Tutorial 2: Peak Detection

Peak detection relies on [`mlgidDETECT`](https://github.com/mlgid-project/mlgidDETECT) package.
First, create the mlgidBASE class instance:



```python
from mlgidbase import mlgidBASE
filename = '../../example/BA2PbI4.h5'
analysis = mlgidBASE(filename=filename)
```

Then run detection:

---
#### **Minimal Code Example**

```python
analysis.run_detection()
```

---

### Parameters

- `entry` (str) — Data file entry to process. Defaults to `None` (all entries). OPTIONAL
- `frame_num` (int/List[int]) — Frame number in each entry for detection / list of frames. Defaults to `None` (all frames).
- `config_file` (str/dict/mlgiddetect.Config/None) — Path to detection configuration YAML file / Dictionary containing configuration parameters. Defaults to `None` (default parameters). OPTIONAL
- `model_type` (str) — Name of the model  `dino` or `faster_rcnn`. OPTIONAL

---

### Decription
The user can specify a single entry or process all entries in the file by setting `entry=None`. The `frame_num` parameter accepts either a single `int` or a `list` of frame indices to process.

The detection configuration can be loaded from a [YAML file](https://github.com/mlgid-project/mlgidDETECT/blob/main/detr.yaml) or set as a dictionary. Additionally, the detection model can be selected via the `model_type` parameter. Currently, two models are supported: `dino` and `faster_rcnn`, as described in the recent work by C. Völter.

Informational logs indicate whether the detection model is being loaded. After the initial load, both the model and its configuration remain fixed; any subsequent configuration inputs will be ignored. To apply a different configuration, a new `mlgidBASE` instance must be created with the desired settings.

Configuration from a YAML file:

```python
analysis = mlgidBASE(filename=filename)
analysis.run_detection(
    entry='entry_0000',
    frame_num=0,
    config_detect='../../example/dino.yaml',
    model_type = 'dino')
print(f"PREPROCESSING_LOG key: {analysis.config_detect.PREPROCESSING_LOG}", '\n',
      f"PREPROCESSING_HISTOGRAMEQUALIZATION key: {analysis.config_detect.PREPROCESSING_HISTOGRAMEQUALIZATION}", '\n',
      f"MODEL_TYPE key: {analysis.config_detect.MODEL_TYPE}")
```

Configuration from as a dictionary:

```python
config_dict = {
    'PREPROCESSING': {
        'LOG': True,
        'HISTOGRAMEQUALIZATION': False
    }
}

analysis = mlgidBASE(filename=filename)
analysis.run_detection(
    entry='entry_0000',
    frame_num=0,
    config_detect=config_dict)
print(f"PREPROCESSING_LOG key: {analysis.config_detect.PREPROCESSING_LOG}", '\n',
      f"PREPROCESSING_HISTOGRAMEQUALIZATION key: {analysis.config_detect.PREPROCESSING_HISTOGRAMEQUALIZATION}", '\n',
      f"MODEL_TYPE key: {analysis.config_detect.MODEL_TYPE}")
```

Load configuration from a YAML file and change the parameters:

```python
import yaml
config_file = '../../example/dino.yaml'
with open(config_file, 'r', encoding='utf-8') as file:
    config_dict = yaml.safe_load(file)

config_dict['PREPROCESSING']['LOG'] = False

analysis = mlgidBASE(filename=filename)
analysis.run_detection(
    entry='entry_0000',
    frame_num=0,
    config_detect=config_dict)

print(f"PREPROCESSING_LOG key: {analysis.config_detect.PREPROCESSING_LOG}", '\n',
      f"PREPROCESSING_HISTOGRAMEQUALIZATION key: {analysis.config_detect.PREPROCESSING_HISTOGRAMEQUALIZATION}", '\n',
      f"MODEL_TYPE key: {analysis.config_detect.MODEL_TYPE}")
```
